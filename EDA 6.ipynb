```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("/content/healthcare_preprocessed (1).csv")
```

```python
plt.figure(figsize=(10,10))
sns.histplot(
    data = df,
    x="Age",
    bins = 25,
    kde=True
    )

plt.title("Distribution of patient age")
plt.xlabel("Age")
plt.ylabel("Number of patients")

plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.histplot(
    data = df,
    x="Age",
    hue = "Gender",
    bins = 15,
    kde=True,
    element="step"
    )

plt.title("Distribution of patient age")
plt.xlabel("Age")
plt.ylabel("Number of patients")

plt.show()
```

```python
gender_counts = df['Gender'].value_counts()

plt.figure(figsize=(8, 6))

sns.barplot(
    x=gender_counts.index,
    y=gender_counts.values
    )

plt.xlabel('Patient distribution by Gender')
plt.ylabel('Gender')
plt.title('Number of patients')
plt.show()
```

```python
condition_counts = df['Medical_Condition'].value_counts()

plt.figure(figsize=(8, 6))

sns.barplot(
    x=condition_counts.index,
    y=condition_counts.values
    )

plt.xlabel('Patient condition distribution')
plt.ylabel('Medical History')
plt.title('Number of patients')
plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.countplot(
    data = df,
    x="Medical_Condition",
    hue = "Gender"
  )
plt.title("Gender Distribution Across Medical Conditions")
plt.xlabel("Medical Condition")
plt.ylabel("Number of Patients")

plt.xticks(rotation=30)
plt.show()
```

```python
admissions_counts = df['Admission_Type'].value_counts()

plt.figure(figsize=(8, 6))

sns.barplot(
    x=admissions_counts.index,
    y=admissions_counts.values
)

plt.xlabel('Patient admission of types')
plt.ylabel('Admission Type')
plt.title('Number of patients')
plt.show()
```

```python
admission_condition = pd.crosstab(
    df['Medical_Condition'],
    df['Admission_Type']
    )
print(admission_condition)
```

```python
admission_condition.plot(
    kind='bar',
    figsize=(10,6),
    stacked=True
)

plt.title("Admission Types Across Medical Conditions")
plt.xlabel("Medical Condition")
plt.ylabel("Number of Patients")

plt.xticks(rotation=30)
plt.legend(title='Admission Type')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.histplot(
    data = df,
    x="Billing_Amount",
    bins = 20,
    kde=True
  )
plt.title("Distribution of Billing Amount")
plt.xlabel("Billing Amount")
plt.ylabel("Number of Patients")

plt.show()
```
